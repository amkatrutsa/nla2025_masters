### Код для работы с файлами .off:

```python
import numpy as np
from pathlib import Path
from typing import List, Optional, Sequence, Tuple, Union, TextIO

class OffMesh:
    """
    Minimal, robust OFF loader/saver and vertex accessor.
    - verts: (N,3) float64 ndarray
    - faces: list[list[int]]
    - header: original header string (e.g. 'OFF' or 'COFF')
    """

    def __init__(self, verts: Optional[np.ndarray] = None, faces: Optional[Sequence[Sequence[int]]] = None, header: str = "OFF"):
        self.verts = np.asarray(verts, dtype=np.float64) if verts is not None else np.zeros((0, 3), dtype=np.float64)
        if self.verts.ndim == 1 and self.verts.size == 3:
            self.verts = self.verts.reshape(1, 3)
        if self.verts.ndim != 2 or self.verts.shape[1] != 3:
            raise ValueError("verts must be of shape (N,3)")
        self.faces = [list(map(int, f)) for f in (faces or [])]
        self.header = header

    # ---------- I/O ----------
    @classmethod
    def load(cls, path_or_file: Union[str, Path, TextIO]) -> "OffMesh":
        """Load OFF from file path or file-like object. Robust to comments and blank lines."""
        need_close = False
        if hasattr(path_or_file, "read"):
            fh = path_or_file  # type: ignore
        else:
            fh = open(path_or_file, "r", encoding="utf-8", errors="replace")
            need_close = True
        try:
            # read and strip lines, ignore comments and empty
            raw_lines = [ln.rstrip("\n") for ln in fh]
            lines = [ln.strip() for ln in raw_lines if ln.strip() and not ln.lstrip().startswith("#")]
            if not lines:
                raise ValueError("Empty or comment-only OFF file")

            # header detection
            first = lines[0].split()
            if first[0].upper().startswith("OFF") or first[0].upper().startswith("COFF") or first[0].upper().startswith("NOFF"):
                header = first[0]
                # if counts are on same line, they are after header tokens
                if len(first) > 1:
                    counts_tokens = first[1:]
                    idx = 1
                else:
                    idx = 1
                    counts_tokens = lines[idx].split()
                    idx += 1
            else:
                header = "OFF"
                counts_tokens = first
                idx = 1

            if len(counts_tokens) < 1:
                raise ValueError("Missing counts line after OFF header")

            nv = int(counts_tokens[0])
            nf = int(counts_tokens[1]) if len(counts_tokens) > 1 else 0

            # read vertices: assume each vertex on its own (robust to extra whitespace)
            verts_list: List[Tuple[float, float, float]] = []
            for vline in lines[idx: idx + nv]:
                toks = vline.split()
                if len(toks) < 3:
                    raise ValueError(f"Malformed vertex line: '{vline}'")
                verts_list.append((float(toks[0]), float(toks[1]), float(toks[2])))
            if len(verts_list) != nv:
                raise ValueError(f"Expected {nv} vertices, found {len(verts_list)}")
            idx += nv

            # read faces
            faces: List[List[int]] = []
            for fline in lines[idx: idx + nf]:
                toks = fline.split()
                if len(toks) < 1:
                    continue
                cnt = int(toks[0])
                if len(toks) - 1 < cnt:
                    # try to gather from following lines (rare)
                    raise ValueError("Face line shorter than count token")
                face_idx = [int(x) for x in toks[1:1 + cnt]]
                faces.append(face_idx)

            verts_arr = np.asarray(verts_list, dtype=np.float64)
            return cls(verts=verts_arr, faces=faces, header=header)
        finally:
            if need_close:
                fh.close()

    def save(self, path: Union[str, Path], float_fmt: str = "{:.6f}") -> None:
        """Write mesh to OFF file. Faces are written as 'n i j k ...' lines."""
        p = Path(path)
        with p.open("w", encoding="utf-8") as fh:
            fh.write(f"{self.header}\n")
            fh.write(f"{self.verts.shape[0]} {len(self.faces)} 0\n")
            for v in self.verts:
                fh.write(" ".join(float_fmt.format(float(x)) for x in v) + "\n")
            for f in self.faces:
                fh.write(str(len(f)) + " " + " ".join(str(int(i)) for i in f) + "\n")

    def get_vertices(self) -> np.ndarray:
        """Return a copy of vetrices (N,3)."""
        return self.verts.copy()

    def set_vertices(self, new_vertices: Union[Sequence[Sequence[float]], np.ndarray]) -> None:
        """Replace vertices. Must match N rows."""
        arr = np.asarray(new_vertices, dtype=np.float64)
        if arr.ndim != 2 or arr.shape[1] != 3:
            raise ValueError("new_vertices must have shape (N,3)")
        if arr.shape[0] != self.verts.shape[0]:
            raise ValueError(f"vertex count mismatch: current {self.verts.shape[0]} vs new {arr.shape[0]}")
        self.verts = arr

    def __repr__(self) -> str:
        return f"OffMesh(verts={self.verts.shape}, faces={len(self.faces)}, header='{self.header}')"
```

```python
import numpy as np
import plotly.graph_objects as go

def visualize_rotate(data,
                     x_eye: float = 1.25,
                     y_eye: float = 1.25,
                     z_eye: float = 0.8,
                     step: float = 0.1,
                     play_button: bool = True) -> go.Figure:
    """
    Build a rotating-camera Plotly Figure for given trace list `data`.
    - data: list of plotly traces (e.g. [go.Mesh3d(...)] or [go.Scatter3d(...)])
    - returns: plotly.graph_objects.Figure (animation frames added, not shown)
    """
    frames = []

    def rotate_z(x, y, z, theta):
        w = x + 1j * y
        w2 = np.exp(1j * theta) * w
        return float(np.real(w2)), float(np.imag(w2)), float(z)

    t_vals = np.arange(0.0, 10.26, step)
    for t in t_vals:
        xe, ye, ze = rotate_z(x_eye, y_eye, z_eye, -t)
        frames.append(dict(layout=dict(scene=dict(camera=dict(eye=dict(x=xe, y=ye, z=ze))))))

    updatemenus = []
    if play_button:
        updatemenus = [dict(type='buttons',
                            showactive=False,
                            y=1,
                            x=0.8,
                            xanchor='left',
                            yanchor='bottom',
                            pad=dict(t=45, r=10),
                            buttons=[dict(label='Play',
                                          method='animate',
                                          args=[None, dict(frame=dict(duration=50, redraw=True),
                                                           transition=dict(duration=0),
                                                           fromcurrent=True,
                                                           mode='immediate')])])]

    fig = go.Figure(data=data,
                    layout=go.Layout(updatemenus=updatemenus),
                    frames=frames)
    # keep aspect correct by data
    fig.update_layout(scene=dict(aspectmode='data'))
    return fig

def visualize_offmesh(offmesh, play_button=True, mesh_color='white', mesh_opacity=1.0, marker_size=2):
    """
    Visualize an OffMesh instance using your rotate animation.
    - offmesh: instance with .get_vertices() -> (N,3) ndarray and .faces -> list[list[int]]
    - returns plotly.Figure and shows it.
    """
    verts = offmesh.get_vertices()
    faces = getattr(offmesh, "faces", None)
    x, y, z = verts.T

    # triangulate polygonal faces to triangle lists for Mesh3d
    if faces:
        i_list, j_list, k_list = [], [], []
        for f in faces:
            if len(f) < 3:
                continue
            if len(f) == 3:
                i_list.append(int(f[0])); j_list.append(int(f[1])); k_list.append(int(f[2]))
            else:
                for t in range(1, len(f) - 1):
                    i_list.append(int(f[0])); j_list.append(int(f[t])); k_list.append(int(f[t+1]))
        if len(i_list) > 0:
            trace = go.Mesh3d(x=x, y=y, z=z, i=np.array(i_list), j=np.array(j_list), k=np.array(k_list),
                              color=mesh_color, opacity=mesh_opacity, flatshading=True)
            fig = visualize_rotate([trace])
        else:
            # fallback to point cloud
            trace = go.Scatter3d(x=x, y=y, z=z, mode='markers')
            fig = visualize_rotate([trace])
            fig.update_traces(marker=dict(size=marker_size,
                                          line=dict(width=2, color='DarkSlateGrey')),
                              selector=dict(mode='markers'))
    else:
        trace = go.Scatter3d(x=x, y=y, z=z, mode='markers')
        fig = visualize_rotate([trace])
        fig.update_traces(marker=dict(size=marker_size,
                                      line=dict(width=2, color='DarkSlateGrey')),
                          selector=dict(mode='markers'))

    fig.update_layout(scene=dict(aspectmode='data'))
    fig.show()
```
